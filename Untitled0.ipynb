```python
!pip install osmnx matplotlib pillow geopandas --quiet

import osmnx as ox
import geopandas as gpd
import shutil
import os
```

```python
place_name = "Atlanta, Georgia, USA"
G = ox.graph_from_place(place_name, network_type='walk')
```

```python
# 2. ArcGIS 호환용 데이터프레임 변환
nodes, edges = ox.graph_to_gdfs(G, nodes=True, edges=True)

```

```python
# 3. 데이터 오류를 유발하는 복잡한 리스트 타입 제거
for col in edges.columns:
    if edges[col].apply(lambda x: isinstance(x, list)).any():
        edges[col] = edges[col].astype(str)


```

```python
# 4. 파일 저장용 폴더 생성 및 Shapefile 내보내기
folder_name = "clean_atlanta_walk"
if os.path.exists(folder_name): shutil.rmtree(folder_name)
os.makedirs(folder_name)
```

```python

# 필수 파일인 엣지(도로 선)만 깔끔하게 내보냅니다.
edges.to_file(os.path.join(folder_name, "atlanta_walk_edges.shp"), encoding='utf-8')

# 5. 한 번에 다운로드할 수 있게 ZIP 압축
shutil.make_archive("clean_atlanta_network", 'zip', folder_name)
print("🎉 성공! 코랩 왼쪽 파일 창에서 [clean_atlanta_network.zip]을 다운로드 후 압축을 풀어두세요!")
```

### Visualize Atlanta Area with Walking Network

```python
# Load the provided shapefile for the Atlanta area (e.g., census blocks)
atlanta_area_gdf = gpd.read_file('/content/tl_2025_13_bg.shp')

# Load the walking network edges from the 'clean_atlanta_walk' folder
atlanta_walk_edges_gdf = gpd.read_file('clean_atlanta_walk/atlanta_walk_edges.shp')

# Print their CRSs to check for compatibility
print(f"CRS for Atlanta area: {atlanta_area_gdf.crs}")
print(f"CRS for Atlanta walk network: {atlanta_walk_edges_gdf.crs}")

# Ensure both GeoDataFrames have the same CRS for plotting
# Reproject if necessary to a common projected CRS like EPSG:3857 (Web Mercator)
# or a local projection for accurate area calculations.
# For simple visualization, let's reproject to the CRS of the walk network if different.
if atlanta_area_gdf.crs != atlanta_walk_edges_gdf.crs:
    print("Reprojecting Atlanta area GeoDataFrame to match walk network CRS...")
    atlanta_area_gdf = atlanta_area_gdf.to_crs(atlanta_walk_edges_gdf.crs)

# Display the first few rows of both GeoDataFrames to confirm loading
print("\nAtlanta Area GeoDataFrame head:")
display(atlanta_area_gdf.head())
print("\nAtlanta Walk Edges GeoDataFrame head:")
display(atlanta_walk_edges_gdf.head())
```

```python
import matplotlib.pyplot as plt

# Create a plot
fig, ax = plt.subplots(figsize=(12, 12))

# Plot the Atlanta area (e.g., census blocks)
atlanta_area_gdf.plot(ax=ax, color='lightgray', edgecolor='white', linewidth=0.5)

# Plot the walking network on top
atlanta_walk_edges_gdf.plot(ax=ax, color='blue', linewidth=0.7, alpha=0.7)

# Set title and labels
ax.set_title('Atlanta Area with Walking Network', fontsize=15)
ax.set_xlabel('Longitude', fontsize=12)
ax.set_ylabel('Latitude', fontsize=12)

# Remove axis ticks for cleaner visualization
ax.set_xticks([])
ax.set_yticks([])

plt.tight_layout()
plt.show()
```
